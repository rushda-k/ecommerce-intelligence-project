```python
import pandas as pd
```

```python
from pathlib import Path

DATA_PATH = Path("../01_Raw_Data/archive (1)")
```

```python
orders = pd.read_csv(
    DATA_PATH / "olist_orders_dataset.csv"
)
```

```python
orders_clean = orders.copy()
```

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in date_columns:
    orders_clean[column] = pd.to_datetime(
        orders_clean[column],
        errors="coerce"
    )
```

```python
orders_clean.dtypes
```

```python
delivered_orders = orders_clean[
    orders_clean["order_status"] == "delivered"
].copy()
```

```python
delivered_orders["delivery_days"] = (
    delivered_orders["order_delivered_customer_date"]
    - delivered_orders["order_purchase_timestamp"]
).dt.total_seconds() / (24 * 60 * 60)
```

```python
delivered_orders["delivery_days"].mean()
```

```python
average_delivery_days = delivered_orders["delivery_days"].mean()

round(average_delivery_days, 2)
```

```python
delivered_orders["is_on_time"] = (
    delivered_orders["order_delivered_customer_date"]
    <= delivered_orders["order_estimated_delivery_date"]
)
```

```python
on_time_rate = delivered_orders["is_on_time"].mean() * 100

round(on_time_rate, 2)
```

```python
late_rate = 100 - on_time_rate

round(late_rate, 2)
```

```python
delivered_orders["days_late"] = (
    delivered_orders["order_delivered_customer_date"]
    - delivered_orders["order_estimated_delivery_date"]
).dt.total_seconds() / (24 * 60 * 60)
```

```python
average_days_late = late_orders["days_late"].mean()

round(average_days_late, 2)
```

```python
late_orders = delivered_orders[
    delivered_orders["days_late"] > 0
].copy()
```

```python
average_days_late = late_orders["days_late"].mean()

round(average_days_late, 2)
```

```python
customers = pd.read_csv(
    DATA_PATH / "olist_customers_dataset.csv"
)
```

```python
customers.head()
```

```python
orders_with_customers = delivered_orders.merge(
    customers[
        [
            "customer_id",
            "customer_city",
            "customer_state"
        ]
    ],
    on="customer_id",
    how="left"
)
```

```python
orders_with_customers[
    [
        "order_id",
        "customer_id",
        "customer_city",
        "customer_state",
        "delivery_days",
        "is_on_time"
    ]
].head()
```

```python
state_delivery = (
    orders_with_customers
    .groupby("customer_state")
    .agg(
        total_orders=("order_id", "count"),
        on_time_rate=("is_on_time", "mean"),
        avg_delivery_days=("delivery_days", "mean")
    )
    .reset_index()
)

state_delivery["on_time_rate"] = (
    state_delivery["on_time_rate"] * 100
)

state_delivery["on_time_rate"] = (
    state_delivery["on_time_rate"].round(2)
)

state_delivery["avg_delivery_days"] = (
    state_delivery["avg_delivery_days"].round(2)
)
```
